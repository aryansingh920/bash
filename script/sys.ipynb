```python
import sys

# Save the original screen output so we can restore it later
original_stdout = sys.stdout

with open('output.log', 'w') as f:
    sys.stdout = f  # Redirect all future prints to the file
    print("This will go into the file, not the screen.")

# Restore the screen
sys.stdout = original_stdout
print("This is back on your terminal!")
```
