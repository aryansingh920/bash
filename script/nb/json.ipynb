```python
import json

# Your dictionary
user_data = {
    "id": 101,
    "name": "Architect",
    "roles": ["admin", "editor"],
    "active": True
}

# Convert to JSON string
json_string = json.dumps(user_data, indent=4)

print(json_string)
# Output: A formatted string with quotes, braces, and indentation.
```

```python
import json
from pathlib import Path

config = {"theme": "dark", "version": "2.1.0"}

# Open a file in write mode and dump the dict into it
with open("config.json", "w") as f:
    json.dump(config, f, indent=4)

print("Dictionary saved to config.json")
```

```python
import json
import sys

user_data = {
    "id": 101,
    "name": "Architect",
    "roles": ["admin", "editor"],
    "active": True
}

# 1. Open the destination file
f = open('output.json', 'w')

# 2. Save the original terminal output
original_stdout = sys.stdout

# 3. Redirect stdout to the file
sys.stdout = f

# 4. Now, printing or dumping goes straight to 'output.json'
# We use json.dump() but tell it to target sys.stdout
json.dump(user_data, sys.stdout, indent=4)

# 5. Restore the terminal connection and close the file
sys.stdout = original_stdout
f.close()

print("This is back on your screen. Check output.json for the data!")
```
