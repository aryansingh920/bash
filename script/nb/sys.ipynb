```python
import subprocess

# Simple command: 'ls -l' (on Mac/Linux) or 'dir' (on Windows)
# Note: We pass the command as a LIST of strings, not one long string.
result = subprocess.run(["ls", "-l"])

print(f"Exit code: {result.returncode}")
# 0 usually means success!
```

```python
# Check which version of git is installed
result = subprocess.run(["git", "--version"], capture_output=True, text=True)

if result.returncode == 0:
    print(f"System has {result.stdout.strip()}")
else:
    print(f"Error: {result.stderr}")
```

```python
# Goal: Run 'ls' and send the output to 'grep' to find .py files
p1 = subprocess.Popen(["ls"], stdout=subprocess.PIPE)
p2 = subprocess.Popen(["grep", ".py"], stdin=p1.stdout,
                      stdout=subprocess.PIPE, text=True)

# Allow p1 to receive a SIGPIPE if p2 exits
p1.stdout.close()

output, error = p2.communicate()
print(output)
```

```python
import subprocess

# 1. Run the first command and capture its output
result_a = subprocess.run(["ls"], capture_output=True, text=True)

# 2. Pass that output directly into the next command's 'input'
result_b = subprocess.run(
    ["grep", ".py"], 
    input=result_a.stdout, # Feed the string here
    capture_output=True, 
    text=True
)

print(result_b.stdout)
```

```python
import requests
import sys

url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    # 1. Send the GET request
    response = requests.get(url)

    # 2. Check if the server "rejected" the request (404, 500, etc.)
    response.raise_for_status()

    # 3. Convert JSON string response into a Python Dictionary
    data = response.json()

    print("--- GET REQUEST SUCCESS ---")
    print(f"Title: {data['title']}")
    print(f"Body: {data['body']}")

except requests.exceptions.HTTPError as err:
    print(f"System Error: {err}")
    sys.exit(1)  # Architect's move: Exit if the connection fails
```

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

# 1. Our data as a Python Dictionary
new_post = {
    "title": "Architect's Manifesto",
    "body": "Do not negotiate with systems that hit the reject button.",
    "userId": 1
}

# 2. Send the POST request
# Using the 'json=' parameter automatically sets the Content-Type to 'application/json'
response = requests.post(url, json=new_post)
print(response)
if response.status_code == 201:  # 201 means "Created"
    print("--- POST REQUEST SUCCESS ---")
    # The server returns the object we created + a new ID
    print(json.dumps(response.json(), indent=4))
else:
    print(f"Failed with status: {response.status_code}")
```
